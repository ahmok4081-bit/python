```python
class Rectangle:
  #create the instuture
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * self.width + 2 * self.height

# Create the object
rectangle1 = Rectangle(3, 4)
```

```python
print(f"The area of the rectangle is: {rectangle1.area()}")
print(f"The perimeter of the rectangle is: {rectangle1.perimeter()}")
```

```python
class Rectangle:
  #create the constructor
  def __init__(self,width,height):
    self.width = width
    self.height = height
  # create the method for the class = action
  def area(self):
    return self.width * self.height
  def perimeter(self):
    return 2 * self.width + 2 * self.height

#create the object
rectangle1 = Rectangle(5, 9)

# Access attributes directly and call methods
print(f"The width is: {rectangle1.width}")
print(f"The height is: {rectangle1.height}")
print(f"The area is: {rectangle1.area()}")
print(f"The perimeter is: {rectangle1.perimeter()}")
```
